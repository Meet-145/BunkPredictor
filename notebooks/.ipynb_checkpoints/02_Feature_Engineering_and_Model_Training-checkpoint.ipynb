# Bunk Predictor — Feature Engineering & ML Model Training

**Goal:** Predict whether a student attends an individual lecture using supervised learning.

This dataset is synthetic, so metrics evaluate simulated patterns, not real-world performance. We use a chronological split and exclude generator-only metadata to avoid leakage.

## 1. Import libraries and load datasets
Update `DATA_DIR` if needed.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, ConfusionMatrixDisplay, roc_auc_score
import joblib

DATA_DIR = "../data"
attendance = pd.read_csv(os.path.join(DATA_DIR, "attendance.csv"), parse_dates=["date"])
students = pd.read_csv(os.path.join(DATA_DIR, "students.csv"))
print("Attendance:", attendance.shape)
print("Students:", students.shape)
attendance.head()
```

## 2. Feature engineering
Prior attendance features use only earlier records for each student, never the current label or future records.

```python
attendance = attendance.sort_values(["student_id", "date", "slot_number"]).copy()
attendance["month"] = attendance["date"].dt.month
attendance["week_of_semester"] = ((attendance["date"] - attendance["date"].min()).dt.days // 7 + 1).astype(int)
attendance["is_friday"] = (attendance["day"] == "Friday").astype(int)
attendance["is_late_lecture"] = (attendance["slot_number"] >= 8).astype(int)
attendance["is_early_lecture"] = (attendance["slot_number"] <= 2).astype(int)
attendance["is_lab"] = (attendance["session_type"] == "Lab").astype(int)
attendance["is_practical"] = (attendance["session_type"] == "Practical/Programming").astype(int)

student_cols = ["student_id", "commute_time_minutes", "previous_semester_attendance_pct", "previous_cgpa_synthetic"]
attendance = attendance.merge(students[student_cols], on="student_id", how="left", validate="many_to_one")

attendance["prior_attendance_rate"] = attendance.groupby("student_id")["present_label"].transform(lambda s: s.shift().expanding().mean())
attendance["prior_attendance_rate"] = attendance["prior_attendance_rate"].fillna(attendance["previous_semester_attendance_pct"] / 100)
attendance["prior_subject_attendance_rate"] = attendance.groupby(["student_id", "subject_code"])["present_label"].transform(lambda s: s.shift().expanding().mean())
attendance["prior_subject_attendance_rate"] = attendance["prior_subject_attendance_rate"].fillna(attendance["previous_semester_attendance_pct"] / 100)

print("Features ready.")
display(attendance[["student_id", "date", "day", "time_slot", "subject_code", "prior_attendance_rate", "prior_subject_attendance_rate", "present_label"]].head(10))
```

## 3. Select features and split chronologically
Earlier dates are used for training and later dates for testing.

```python
categorical_features = ["day", "time_slot", "subject_code", "session_type"]
numeric_features = ["slot_number", "month", "week_of_semester", "is_friday", "is_late_lecture", "is_early_lecture", "is_lab", "is_practical", "commute_time_minutes", "previous_semester_attendance_pct", "previous_cgpa_synthetic", "prior_attendance_rate", "prior_subject_attendance_rate"]
feature_columns = categorical_features + numeric_features

unique_dates = np.sort(attendance["date"].dropna().unique())
cut_date = unique_dates[int(len(unique_dates) * 0.8)]
train_df = attendance[attendance["date"] < cut_date].copy()
test_df = attendance[attendance["date"] >= cut_date].copy()
X_train, y_train = train_df[feature_columns], train_df["present_label"].astype(int)
X_test, y_test = test_df[feature_columns], test_df["present_label"].astype(int)

print("Cutoff:", pd.Timestamp(cut_date).date())
print("Train rows:", len(train_df), "| Test rows:", len(test_df))
print("Train attendance rate:", round(y_train.mean(), 3))
print("Test attendance rate:", round(y_test.mean(), 3))
```

## 4. Train Logistic Regression and Random Forest
Preprocessing is fit only on training data through a pipeline.

```python
numeric_preprocessor = Pipeline([("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler())])
categorical_preprocessor = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")), ("onehot", OneHotEncoder(handle_unknown="ignore"))])
preprocessor = ColumnTransformer([("num", numeric_preprocessor, numeric_features), ("cat", categorical_preprocessor, categorical_features)])

logistic_model = Pipeline([("preprocessor", preprocessor), ("classifier", LogisticRegression(max_iter=1000, class_weight="balanced"))])
random_forest_model = Pipeline([("preprocessor", preprocessor), ("classifier", RandomForestClassifier(n_estimators=200, min_samples_leaf=5, class_weight="balanced", random_state=42, n_jobs=-1))])

logistic_model.fit(X_train, y_train)
random_forest_model.fit(X_train, y_train)
print("Both models trained.")
```

## 5. Evaluate both models
Compare accuracy, precision, recall, F1 and ROC-AUC. Positive class = Present.

```python
def evaluate_model(name, model):
    pred = model.predict(X_test)
    proba = model.predict_proba(X_test)[:, 1]
    return {"Model": name, "Accuracy": accuracy_score(y_test, pred),
            "Precision (Present)": precision_score(y_test, pred, zero_division=0),
            "Recall (Present)": recall_score(y_test, pred, zero_division=0),
            "F1 (Present)": f1_score(y_test, pred, zero_division=0),
            "ROC-AUC": roc_auc_score(y_test, proba) if y_test.nunique() == 2 else np.nan}

results = pd.DataFrame([evaluate_model("Logistic Regression", logistic_model), evaluate_model("Random Forest", random_forest_model)]).set_index("Model")
display(results.round(3))
```

```python
models = {"Logistic Regression": logistic_model, "Random Forest": random_forest_model}
best_name = results["F1 (Present)"].idxmax()
best_model = models[best_name]
predictions = best_model.predict(X_test)

print("Best baseline by F1:", best_name)
print(classification_report(y_test, predictions, labels=[0, 1], target_names=["Absent", "Present"], zero_division=0))
ConfusionMatrixDisplay.from_predictions(y_test, predictions, display_labels=["Absent", "Present"], values_format="d")
plt.title(f"Confusion Matrix — {best_name}")
plt.tight_layout()
plt.show()
```

## 6. Save the model
The saved pipeline includes preprocessing and the classifier.

```python
MODEL_DIR = "../models"
os.makedirs(MODEL_DIR, exist_ok=True)
joblib.dump(best_model, os.path.join(MODEL_DIR, "attendance_model.pkl"))
joblib.dump(feature_columns, os.path.join(MODEL_DIR, "attendance_feature_columns.pkl"))
print("Saved model and feature list in:", os.path.abspath(MODEL_DIR))
```

## 7. Sample prediction
This uses one held-out test row only to demonstrate the prediction interface.

```python
sample = X_test.iloc[[0]]
prob_present = best_model.predict_proba(sample)[0, 1]
predicted_label = int(best_model.predict(sample)[0])
display(sample)
print(f"Predicted probability of attending: {prob_present:.1%}")
print("Predicted outcome:", "Present" if predicted_label else "Absent")
```

## Important notes
- Synthetic data and assumed calendar dates mean these metrics are only a pipeline demonstration.
- Do not include `synthetic_attendance_persona` or `latent_base_attendance_probability` as features; they leak generator metadata.
- The model predicts lecture attendance. The maximum safe bunk count must be calculated mathematically from attended/total lectures and the required threshold.
- Before deployment, validate with real, consented attendance data and review which personal features are truly necessary.
